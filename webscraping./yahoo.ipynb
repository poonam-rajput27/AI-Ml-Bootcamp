```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from webdriver_manager.chrome import ChromeDriverManager
import time
import pandas as pd
```

<div style="padding: 15px 20px; background: linear-gradient(135deg, #fbf9ff 0%, #f5f0ff 100%); border-left: 5px solid #b19ffb; border-radius: 8px; border: 1px solid rgba(177, 159, 251, 0.15); font-family: 'Segoe UI', system-ui, sans-serif; text-align: left; display: flex; align-items: center; gap: 10px;"><span style="font-size: 1.2rem;">⚙️</span><span style="color: #2b2156; font-size: 1.1rem; font-weight: 700; letter-spacing: -0.3px;">Yahoo Finance <span style="color: #7c5cff;">Scraper Function</span></span></div>

```python
def scrape_yahoo_finance():

    options = Options()
    options.add_argument("--headless=new")
    options.add_argument("--disable-gpu")
    options.add_argument("--window-size=1920,1080")

    driver = webdriver.Chrome(
        service=Service(ChromeDriverManager().install()),
        options=options
    )

    url = "https://finance.yahoo.com/markets/stocks/most-active/"
    driver.get(url)

    time.sleep(5)

    stocks = []

    rows = driver.find_elements(By.CSS_SELECTOR, "table tbody tr")

    for row in rows[:20]:

        try:
            symbol = row.find_element(By.CSS_SELECTOR, "td:nth-child(1)").text
        except:
            symbol = ""

        try:
            company = row.find_element(By.CSS_SELECTOR, "td:nth-child(2)").text
        except:
            company = ""

        try:
            price = row.find_element(By.CSS_SELECTOR, "td:nth-child(4)").text
        except:
            price = ""

        try:
            change = row.find_element(By.CSS_SELECTOR, "td:nth-child(5)").text
        except:
            change = ""

        try:
            volume = row.find_element(By.CSS_SELECTOR, "td:nth-child(7)").text
        except:
            volume = ""

        stocks.append({
            "symbol": symbol,
            "company": company,
            "price": price,
            "change": change,
            "volume": volume
        })

    driver.quit()

    return stocks
```

<div style="padding: 15px 20px; background: linear-gradient(135deg, #fbf9ff 0%, #f5f0ff 100%); border-left: 5px solid #b19ffb; border-radius: 8px; border: 1px solid rgba(177, 159, 251, 0.15); font-family: 'Segoe UI', system-ui, sans-serif; text-align: left; display: flex; align-items: center; gap: 10px;"><span style="font-size: 1.2rem;">🚀</span><span style="color: #2b2156; font-size: 1.1rem; font-weight: 700; letter-spacing: -0.3px;">Executing <span style="color: #7c5cff;">Scraper Pipeline</span></span></div>

```python
 stocks_data=scrape_yahoo_finance()
```

```python
print(stocks_data)
```

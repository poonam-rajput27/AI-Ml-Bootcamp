```python
BASE_URL = "https://en.wikipedia.org/w/api.php"
HEADERS = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36"
}
```

<div style="padding: 15px 20px; background: linear-gradient(135deg, #f4f7fa 0%, #e9eff5 100%); border-left: 5px solid #234e8c; border-radius: 8px; border: 1px solid rgba(35, 78, 140, 0.2); font-family: 'Segoe UI', system-ui, sans-serif; text-align: left; display: flex; align-items: center; gap: 10px;">

    <span style="color: #162a45; font-size: 1.1rem; font-weight: 700; letter-spacing: -0.3px;">
     Featch <span style="color: #234e8c; font-style: italic;">Data </span>
    </span>
</div>

```python
#Fetch search data directly (getting 20 results)
search_payload = {
    "action": "query",
    "list": "search",
    "srsearch": "technology", 
    "format": "json",
    "utf8": 1,
    "srlimit": 20
}
response = requests.get(BASE_URL, params=search_payload, headers=HEADERS, timeout=20)
search_results = response.json().get("query", {}).get("search", [])

articles = []
```

```python
# 2. Extract baseline details without fetching the summary text
for item in search_results:
    page_id = item.get("pageid")
    
    articles.append({
        "title": item.get("title"),
        "page_id": page_id,
        "word_count": item.get("wordcount"),
        "category": "Wikipedia Article",
        "link": f"https://en.wikipedia.org/?curid={page_id}"
    })
```

```python
df = pd.DataFrame(articles)

# Setup layout options for display
pd.set_option('display.max_columns', None)
pd.set_option('display.width', 1000)

# Display the clean dataset table layout
df
```

```python
import requests
from bs4 import BeautifulSoup
url="https://www.amazon.in/s?k=laptop&crid=YLH7SZQM1WE6&sprefix=laptop%2Caps%2C610&ref=nb_sb_noss_2"
headers = {
'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/109.0.0.0 Safari/537.36',
'Accept': 'text/html,application/xhtml+xml,application/xml;q=0.9,image/avif,image/webp,*/*;q=0.8',
'Accept-Language': 'en-US,en;q=0.9',
'Accept-Encoding': 'gzip, deflate, br',
'Referer': 'https://www.google.com/'
}

response = requests.get(url,headers=headers)


```

<h2 style="color:white; margin-top:0;">
 Extracting Data
</h2>

<p style="color:white;text-align:centre; font-size:15px; margin-bottom:0;">
Relevant data is extracted from the parsed HTML structure using predefined rules such as tags, classes, and attributes.
</p>

</div>

```python
laptops = soup.find_all('div',class_='sg-col-inner')
print(laptops) 
```

```python
laptop_data=[]
```

```python
for laptop in laptops :
    title=laptop.find('h2',class_='a-size-medium a-spacing-none a-color-base a-text-normal')
    price= laptop.find('span',class_='a-price-whole')
    rating=laptop.find('span',class_='a-size-small a-color-base')
    
    laptop_data.append(
        {
            'Title':title,
            'price':price,
            'rating':rating,
        }
        )
```

```python
print(laptop_data)
```

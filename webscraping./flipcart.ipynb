```python
import requests
```

```python
from bs4 import BeautifulSoup
url="https://www.flipkart.com/search?q=shoes&as=on&as-show=on&otracker=AS_Query_TrendingAutoSuggest_2_0_na_na_na&otracker1=AS_Query_TrendingAutoSuggest_2_0_na_na_na&as-pos=2&as-type=TRENDING&suggestionId=shoes&requestId=137bfca7-eb1c-4aa8-8ce9-79a525bf0eb2"
headers = {
'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/109.0.0.0 Safari/537.36',
'Accept': 'text/html,application/xhtml+xml,application/xml;q=0.9,image/avif,image/webp,*/*;q=0.8',
'Accept-Language': 'en-US,en;q=0.9',
'Accept-Encoding': 'gzip, deflate, br',
'Referer': 'https://www.google.com/'
}
response=requests.get(url , headers=headers)
```

```python
response.status_code
```

```python
soup=BeautifulSoup(response.content,"html.parser")
soup
```

```python
print(soup.prettify())
```

```python
shoes= soup.find_all('div',class_='bLCLBY nr15la')
print (shoes)
```

```python
shoes_data=[]
```

```python
for shoe in shoes :
    shoe_name=shoe.find('div',class_='Fo1I0b')
    price= shoe.find('div',class_='hZ3P6w')
    size=shoe.find('div',class_='QiMO5r')
    
    shoes_data.append(
        {
            'shoe_name':shoe_name.get_text(strip=True)
            if shoe_name
            else None,
            'price':price.get_text(strip=True)
            if price
            else None
            ,
            'Size':size.get_text(strip=True)
            if size
            else None
            
        }
        )
```

```python
print(shoes_data)
```

```python
import pandas as pd 
```

```python
shoes=pd.DataFrame(shoes_data)
```

```python
print(shoes)
```

```python
BASE_URL = "https://dummyjson.com/products"
```

<div style="padding: 22px 20px; background: linear-gradient(135deg, #244263 0%, #3a5874 100%); border-radius: 10px; font-family: 'Segoe UI', Arial, sans-serif; text-align: center; color: #ffffff; box-shadow: 0 4px 10px rgba(0,0,0,0.12);">
<h2 style="color: #ffffff; margin: 0 0 6px 0; font-size: 26px; font-weight: 700; letter-spacing: -0.3px;">Fetching JSON Data</h2>
<p style="color: #e2e8f0; margin: 0; font-size: 14px; font-weight: 400; opacity: 0.95;">Extract the raw JSON response payload containing product catalog records.</p>
</div>

```python


def fetch_products():
    try:
        # API request
        response = requests.get(BASE_URL, timeout=30)

        # Status check
        if response.status_code != 200:
            return {
                "success": False,
                "error": f"API failed with status {response.status_code}"
            }

        # Convert raw response data into JSON
        data = response.json()
        
        # DummyJSON wraps its list inside a "products" key
        product_list = data.get("products", [])

        products = []

        # Sirf first 5 products lena (matching your assignment logic)
        for item in product_list[:30]:
            products.append({
                "title": item.get("title"),
                "price": item.get("price"),
                "category": item.get("category"),
                "rating": item.get("rating"),
                "image": item.get("thumbnail") # DummyJSON uses 'thumbnail' for the main image url
            })

        return {
            "success": True,
            "products": products
        }

    except requests.exceptions.RequestException as e:
        return {
            "success": False,
            "error": str(e)
        }


# Run test
```

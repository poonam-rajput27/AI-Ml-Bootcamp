```python
BASE_URL = "https://fakestoreapi.com/products"
```

<div style="padding: 22px 20px; background: linear-gradient(135deg, #244263 0%, #3a5874 100%); border-radius: 10px; font-family: 'Segoe UI', Arial, sans-serif; text-align: center; color: #ffffff; box-shadow: 0 4px 10px rgba(0,0,0,0.12);">
  <h2 style="color: #ffffff; margin: 0 0 6px 0; font-size: 26px; font-weight: 700; letter-spacing: -0.3px;">Fetching fakestore Data</h2>
  <p style="color: #e2e8f0; margin: 0; font-size: 14px; font-weight: 400; opacity: 0.95;">Retrieving real-time product catalogs and store inventory via the API endpoint.</p>
</div>

```python

def fetch_products():
    try:
        response = requests.get(BASE_URL, timeout=30)

        if response.status_code != 200:
            return {
                "success": False,
                "error": f"API failed with status {response.status_code}"
            }

        data = response.json()

        products = []

    
        for item in random.sample(data, 20):
            products.append({
                "title": item.get("title"),
                "price": item.get("price"),
                "category": item.get("category"),
                "rating": item.get("rating", {}).get("rate"),
                "image": item.get("image")
            })

        return {
            "success": True,
            "products": products
        }

    except requests.exceptions.RequestException as e:
        return {
            "success": False,
            "error": str(e)
        }

if __name__ == "__main__":
    result = fetch_products()
    print(result)
```

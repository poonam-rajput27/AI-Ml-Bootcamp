```python
import numpy as np
import pandas as pd

np.random.seed(45)
n_student= 10000
data={
    "student_Id" : [f"STU{i:05d}"for i in range (1,n_student +1)],
    "Gender" : np.random.choice(["Male","Female"],n_student),
    "Age" : np.random.randint(18,25,n_student),
    "department" : np.random.choice(["CSE","ME","ECE"],n_student),
    "Residence" : np.random.choice(["Hostler","Day Scholar"], n_student, p=[0.4,0.6]),
    "Family_income" : np.random.randint(2000,200000,n_student),
    "Attendence_percentage" : np.clip(np.random.normal(75,15,n_student),20,100).round(1),
    #"Semester_GPA":np.random.normal(1,10,n_student).round(1),
    "Semester_GPA" : np.clip(np.random.normal(7.0,1.5,n_student),0,10).round(2),
    "Internal_assesment" : np.random.randint(10,30,n_student),
    "backloges" : np.random.randint(0.5,n_student),
    #"assignment_submission_rate":np.random.randint(1,6,n_student),
    "assignment_submission_rate" : np.clip(np.random.normal(85,15,n_student),0,100).round(1),
   # "previous_exam_performance":np.random.choice(["poor","Average","Excellenct"],n_student,p=[0.30,0.60,0.90]),
    "LMS_Login_frequency" : np.random.poisson(0.5,n_student),
    "Libraray_visits" : np.random.poisson(0.5,n_student),
    "club_participation" : np.random.choice([0,1],n_student,p=[0.7,0.3]),
   # "counselling_Session_Attendence":np.random.choice([0,1,2],n_student,p=[0.70,0.20,0.08,0.02]),
    # "participation_in_Events" : np.random.choice(["YES","NO"],n_student),
    "Scholarship_Status" : np.random.choice(["YES","NO"],n_student,p=[0.25,0.75]),
    "Tuition_Fee_Due" : np.random.choice([0,1],n_student,p=[0.8,0.2]),
    #"Education_loan" : np.random.choice(["yes","no"],n_student,p=[0.2,0.8]),
    #"Financial_aid_Recieved" : np.random.choice(["yes","no"],n_student,p=[0.3,0.7]),
    #"student_Attendence_drop" : np.random.choice(["yes","no"],n_student,p=[0.2,0.8]),
    "consecutive_Absence_Days":np.random.randint(1,5, n_student),
    #"Diciplinary_action" : np.random.choice(["yes","no"],n_student),
    "Acadmice_Warnings":np.random.choice([0,1,2,3],n_student,p=[0.7,0.2,0.08,0.02])
    
    

} 
df = pd.DataFrame(data)
print(df)
```

```python
# risk score based on domain knowledge
# calculated weifhted Risk Score
risk_score = (
    0.30 * ( df["Attendence_percentage"] < 60).astype(int) +
    0.25 * ( df["Semester_GPA"] < 5.5).astype( int) +
    0.15 * df["Tuition_Fee_Due"] +
    0.10 * ( df["Acadmice_Warnings"] < 0).astype( int) +
    0.10 * (df["consecutive_Absence_Days"] > 10).astype(int) +
    0.20 * ( df["backloges"] >= 4).astype( int) -
    0.05 * df["club_participation"]- 
    0.05 * ( df["LMS_Login_frequency"] > 25).astype(int)
)

# adding random variation
risk_score += np.random.normal(0,0.05, n_student)

df["Dropout"] = (risk_score >= 0.05).astype(int)
print(df)

df.to_csv("student_dropout_stnthetic_data", index = False)
```
